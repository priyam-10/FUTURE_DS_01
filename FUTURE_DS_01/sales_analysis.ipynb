## KPI Insights

The business generated substantial revenue and profit during the analyzed period. The number of unique customers and orders indicates a diverse customer base and active transaction volume. These KPIs provide a foundation for evaluating business performance across regions, categories, and products.

# Revenue Trend Analysis

```python
df['YearMonth'] = df['order date'].dt.to_period('M')

monthly_sales = df.groupby('YearMonth')['sales'].sum()

plt.figure(figsize=(12,5))
monthly_sales.plot(marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

## Revenue Trend Insights

Monthly sales analysis reveals fluctuations throughout the year, indicating periods of higher and lower demand. Understanding these trends helps the business optimize inventory planning, marketing campaigns, and resource allocation during peak periods.

# Regional Performance

```python
region_sales = df.groupby('region')['sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=region_sales.values, y=region_sales.index )
plt.title('Sales by Region')
plt.show()
```

## Regional Performance Insights

The West region generated the highest sales revenue among all regions, while the South region recorded the lowest sales.

This suggests stronger market penetration and customer demand in the West region.

# Category Analysis

```python
category_sales = df.groupby('category')['sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=category_sales.index, y=category_sales.values)
plt.title('Sales by Category')
plt.show()
```

## Category Analysis Insights

Category-level analysis helps identify the product groups contributing most to revenue. High-performing categories represent key growth opportunities, while lower-performing categories may require strategic review.

# Profit Analysis

```python
category_profit = df.groupby('category')['profit'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(x=category_profit.index, y=category_profit.values)
plt.title('Profit by Category')
plt.show()
```

# Top Products

```python
top_products = (
    df.groupby('product name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
sns.barplot(x=top_products.values, y=top_products.index)
plt.title('Top 10 Products by Sales')
plt.show()
```

## Product Performance Insights

Top-selling products contributed significantly to total revenue and should remain a business focus.

# Sales vs Profit Analysis

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='sales', y='profit')
plt.title('Sales vs Profit')
plt.show()
```

# Top 10 Customers

```python
top_customers = (
    df.groupby('customer name')['sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
sns.barplot(
    x=top_customers.values,
    y=top_customers.index
)

plt.title('Top 10 Customers by Sales')
plt.show()
```

# Segment Analysis

```python
segment_sales = df.groupby('segment')['sales'].sum()

plt.figure(figsize=(8,5))
sns.barplot(
    x=segment_sales.index,
    y=segment_sales.values
)

plt.title('Sales by Segment')
plt.show()
```
